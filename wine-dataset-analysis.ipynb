```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
```

```python
wine=load_wine()
df= pd.DataFrame(wine.data,columns=wine.feature_names)
df["cultiver"]=wine.target_names[wine.target]
df.head()#head starting ke 5 coloum show karta hai
```

```python
print("Shape(row,columns):",df.shape)#show karne ke liye
print("Number of records:",len(df))#total data set
print("\nColumns names:",list(df.columns))
print("\nData type:\n",df.dtypes)#
df.info()
df.tail()
```

```python
print("Missing values per columns:\n",df.isnull().sum())
print("total missing values:",df.isnull().sum().sum())
print("Dupicates rows:",df.duplicated().sum())
df=df.drop_duplicates()#delete
df=df.reset_index(drop=True)#0 se numuring start kerne ke liye
print("Shapes After cleaning:",df.shape)
```

```python
df.describe().round(2)#its display ,mean,std,min,max,skew print karega record ka
```

```python
print("Wines per cultiver:\n", df["cultiver"].value_counts())
print("\nPercentage per cultiver:\n", (df["cultiver"].value_counts(normalize=True)*100).round(1))
```

```python
num_df = df.drop(columns="cultiver")                  # keep only the numeric columns for calculations
stats = pd.DataFrame({                                # build a custom summary table
    "Mean": num_df.mean(),                            # average value of each column
    "Median": num_df.median(),                        # middle value of each column
    "Std Dev": num_df.std(),                          # how spread out the values are
    "Min": num_df.min(),                              # smallest value
    "Max": num_df.max(),                              # largest value
    "Skewness": num_df.skew()                         # positive = long tail on the right, negative = long tail on the left
})
stats.round(2)
```

```python
Q1 = num_df.quantile(0.25)                            # 25th percentile of each column
Q3 = num_df.quantile(0.75)                            # 75th percentile of each column
IQR = Q3 - Q1                                         # Inter-Quartile Range = spread of the middle 50% of data
lower = Q1 - 1.5 * IQR                                # values below this are outliers
upper = Q3 + 1.5 * IQR                                # values above this are outliers
outliers = ((num_df < lower) | (num_df > upper)).sum()  # count outliers in each column
print("Number of outliers per column:")
print(outliers[outliers > 0])
```

```python
group_mean = df.groupby("cultiver").mean().round(2)
group_mean.T
```

```python
print("Cultivar with the highest average alcohol:", group_mean["alcohol"].idxmax())      # which cultivar has the most alcohol
print("Cultivar with the highest average flavanoids:", group_mean["flavanoids"].idxmax())  # which cultivar has the most flavanoids
print("Cultivar with the highest average colour intensity:", group_mean["color_intensity"].idxmax())  # which has the darkest colour
print("Cultivar with the highest average proline:", group_mean["proline"].idxmax())      # which has the most proline
```

```python
df.groupby("cultiver")["alcohol"].agg(["count", "mean", "median", "min", "max"]).round(2)  # several statistics of alcohol per cultivar
```

```python
corr = num_df.corr()                                  # correlation between every pair of columns (-1 to +1)
pairs = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool)).stack().dropna()  # keep each pair only once (upper triangle) and drop empty cells
pairs = pairs.sort_values(ascending=False)            # sort from strongest positive to strongest negative
print("Top 5 positive correlations:\n", pairs.head(5).round(2))   # features that increase together
print("\nTop 5 negative correlations:\n", pairs.tail(5).round(2))  # features where one goes up as the other goes down
```

```python
# --- Graph 1: Bar chart -> number of wines in each cultivar ---
df["cultiver"].value_counts().sort_index().plot(kind="bar", color=["#4C72B0", "#55A868", "#C44E52"])  # one bar per cultivar
plt.title("Bar Chart: Wines per Cultivar")            # graph title
plt.xlabel("Cultivar")                                # x-axis label
plt.ylabel("Number of Wines")                         # y-axis label
plt.xticks(rotation=0)                                # keep names horizontal
plt.show()
```

```python
# --- Graph 2: Pie chart -> share of each cultivar ---
counts = df["cultiver"].value_counts().sort_index()   # count of wines per cultivar
plt.pie(counts.values, labels=counts.index, autopct="%1.1f%%", colors=["#4C72B0", "#55A868", "#C44E52"])  # slices with percentages
plt.title("Pie Chart: Share of Each Cultiver")
plt.show()
```

```python
# --- Graph 3: Histogram -> distribution of alcohol ---
plt.hist(df["alcohol"], bins=15, color="skyblue", edgecolor="black")  # split alcohol values into 15 bins
plt.axvline(df["alcohol"].mean(), color="red", linestyle="--", label="Mean")  # dashed line at the average
plt.title("Histogram: Alcohol Content")
plt.xlabel("Alcohol (%)")
plt.ylabel("Number of Wines")
plt.legend()
plt.show()
```

```python
# --- Graph 4: Box plot -> compare alcohol across cultivars and spot outliers ---
cultivers = sorted(df["cultiver"].unique())           # names of the 3 cultivars in order
data_box = [df[df["cultiver"] == c]["alcohol"] for c in cultivers]  # alcohol values of each cultivar
plt.boxplot(data_box)                                 # box = middle 50%, line = median, dots = outliers
plt.xticks([1, 2, 3], cultivers)                      # name the three boxes
plt.title("Box Plot: Alcohol by Cultiver")
plt.ylabel("Alcohol (%)")
plt.show()
```

```python
# --- Graph 5: Grouped bar chart -> average of selected features per cultivar ---
group_mean[["alcohol", "malic_acid", "magnesium"]].plot(kind="bar", figsize=(8, 5))  # compare 3 features across cultivars
plt.title("Average Alcohol, Malic Acid and Magnesium by Cultivar")
plt.ylabel("Average value")
plt.xticks(rotation=0)
plt.show()
```

```python
# --- Graph 6: Scatter plot -> flavanoids vs total phenols, coloured by cultivar ---
colors = {"class_0": "#4C72B0", "class_1": "#55A868", "class_2": "#C44E52"}  # one colour per cultivar
for c in cultivers:                                   # loop through each cultivar
    part = df[df["cultiver"] == c]                    # rows of that cultivar only
    plt.scatter(part["flavanoids"], part["total_phenols"], c=colors[c], label=c, alpha=0.7)  # plot its points
plt.title("Scatter Plot: Flavanoids vs Total Phenols")
plt.xlabel("Flavanoids")
plt.ylabel("Total Phenols")
plt.legend()                                          # show which colour is which cultivar
plt.show()
```

```python
# --- Graph 7: Line plot -> average proline for each cultivar ---
plt.plot(group_mean.index, group_mean["proline"], marker="o", color="purple")  # line joining the 3 averages
plt.title("Line Plot: Average Proline by Cultiver")
plt.xlabel("Cultiver")
plt.ylabel("Average Proline")
plt.grid(True)                                        # light grid lines for easy reading
plt.show()
```

```python
# --- Graph 8: Correlation heatmap (matplotlib only, no seaborn) ---
n = len(corr)                                         # number of numeric columns
plt.figure(figsize=(10, 8))                           # make the figure bigger (13 columns)
plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)    # draw the matrix as coloured squares
plt.colorbar()                                        # colour scale legend
plt.xticks(range(n), corr.columns, rotation=90)       # column names on x-axis
plt.yticks(range(n), corr.columns)                    # column names on y-axis
for i in range(n):                                    # loop over rows
    for j in range(n):                                # loop over columns
        plt.text(j, i, round(corr.iloc[i, j], 1), ha="center", va="center", fontsize=6)  # write value in each square
plt.title("Correlation Heatmap")
plt.tight_layout()                                    # stop labels from overlapping
plt.show()
```

```python
top_pair = pairs.index[0]                             # the two features with the strongest positive correlation
bottom_pair = pairs.index[-1]                         # the two features with the strongest negative correlation
print("1. The dataset has", len(df), "wines and", num_df.shape[1], "chemical measurements, with", df.isnull().sum().sum(), "missing values.")
print("2. Most common cultiver:", df["cultiver"].value_counts().idxmax(), "with", df["cultiver"].value_counts().max(), "wines.")
print("3. Average alcohol content is", round(df["alcohol"].mean(), 2), "% (range", df["alcohol"].min(), "to", df["alcohol"].max(), ").")
print("4. Highest alcohol on average:", group_mean["alcohol"].idxmax(), "| Lowest:", group_mean["alcohol"].idxmin())
print("5. Strongest positive relationship:", top_pair, "(correlation =", round(pairs.iloc[0], 2), ")")
print("6. Strongest negative relationship:", bottom_pair, "(correlation =", round(pairs.iloc[-1], 2), ")")
print("7. Columns with the most outliers:", list(outliers.sort_values(ascending=False).head(3).index))
```
